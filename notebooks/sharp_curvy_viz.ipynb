# Sharp vs. Curvy


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
import sys
```

```python
def load_sharp_curvy_data(folder_path):
    sharp_curvy_dict = {}
    for file in os.listdir(folder_path):
        if file.endswith('.csv') and 'sharp' in file:
            df = pd.read_csv(os.path.join(folder_path, file))
            sharp_curvy_dict[file] = df
    return sharp_curvy_dict


sharp_curvy_dict = load_sharp_curvy_data("../results/")
```
